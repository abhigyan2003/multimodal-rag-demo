```python
import base64
import os
from langchain_core.messages import HumanMessage
```

```python
def encode_image(image_path):
    """Getting the base64 string"""
    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode("utf-8")


```

```python
from langchain_openai import ChatOpenAI
from langchain_core.messages import HumanMessage

def image_summarize(img_base64, prompt):
    chat = ChatOpenAI(model="gpt-4o", max_tokens=1024)

    msg = chat.invoke(
        [
            HumanMessage(
                content=[
                    {"type": "text", "text": prompt},
                    {
                        "type": "image_url",
                        "image_url": {
                            "url": f"data:image/jpeg;base64,{img_base64}"
                        },
                    },
                ]
            )
        ]
    )
    return msg.content
```

```python
def generate_img_summaries(path):
    """
    Generate summaries and base64 encoded strings for images
    path: Path to list of .jpg files extracted by Unstructured
    """

    # Store base64 encoded images
    img_base64_list = []

    # Store image summaries
    image_summaries = []

    # Prompt
    prompt = """You are an assistant tasked with summarizing images for retrieval.
These summaries will be embedded and used to retrieve the raw image.
Give a concise summary of the image that is well optimized for retrieval.
"""

    base64_image = encode_image(path)
    img_base64_list.append(base64_image)
    image_summaries.append(image_summarize(base64_image, prompt))

    return img_base64_list, image_summaries
```

```python
import google.generativeai as genai

# Load your Gemini API key (replace YOUR_API_KEY with your actual key or load from environment)
genai.configure(api_key="YOUR_API_KEY")
```

```python
import base64
from PIL import Image
import io

def encode_image(image_path):
    """Return image content bytes (not base64) for Gemini API"""
    with open(image_path, "rb") as img_file:
        return img_file.read()
```

```python
def image_summarize_gemini(image_bytes, prompt):
    model = genai.GenerativeModel('gemini-pro-vision')

    response = model.generate_content(
        [prompt, image_bytes],
        generation_config={"max_output_tokens": 1024}
    )

    return response.text
```

```python
def generate_img_summaries(path):
    """
    Generate image summaries using Gemini API
    path: Path to .jpg file
    """

    # Prompt
    prompt = """You are an assistant tasked with summarizing images for retrieval.
These summaries will be embedded and used to retrieve the raw image.
Give a concise summary of the image that is well optimized for retrieval.
"""

    # Get image bytes
    image_bytes = encode_image(path)

    # Summarize using Gemini
    image_summary = image_summarize_gemini(image_bytes, prompt)

    return image_bytes, image_summary
```
